# Initial training run

The goal of the study is to determine that soft labels from a teacher can carry more information that hard labels. This inital experimetn aims to establish a distillation baseline. Training a strong MLP model (parameters [512, 512]) and using that to train a weak one down the line ([128, 128]), we establish training of the soft model on the larger one.

We see that there is a considerable difference between training on the soft labels versus training on the hard labels. In this case, the baseline outperforms the distillation trained on soft labels. In order to account for variation, we train with several seeds and establish 95% confidence interval error bars.

We see that there is a discrepancy of arund 0.015 and this can be due to several reasons. This could be due to sufficient capacity of hte model (i.e. not enough parameters to fully capture complexity). We can try running distillation with different model sizes to capture this quality.
